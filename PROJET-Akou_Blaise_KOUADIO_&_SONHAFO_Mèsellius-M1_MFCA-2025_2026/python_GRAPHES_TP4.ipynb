```python
import numpy as np
from scipy import stats

import matplotlib.pyplot as plt

# Data for alpha = 0 (excluding inf values)
N0 = np.array([16, 32, 64, 128, 256, 512, 1024, 2048, 4096, 8192, 16384])
err0 = np.array([4.005902e-01, 3.656039e-01, 1.230833e+00, 8.573733e-01, 
                 1.741133e+00, 1.725475e-02, 2.458594e-01, 3.221928e-02, 
                 3.867244e-01, 1.183664e+00, 5.186072e-01])

# Data for alpha = 1
N1 = np.array([4, 8, 16, 32, 64, 128, 256, 512, 1024, 2048, 4096, 8192, 16384])
err1 = np.array([2.266187e-01, 5.165156e-02, 1.262676e-02, 3.139192e-03, 
                 7.837101e-04, 1.958596e-04, 4.896067e-05, 1.223990e-05, 
                 3.059966e-06, 7.649950e-07, 1.913014e-07, 4.788688e-08, 
                 1.198609e-08])

# Calculate logarithms
log_N0 = np.log(N0)
log_err0 = np.log(err0)
log_N1 = np.log(N1)
log_err1 = np.log(err1)

# Linear regression
slope0, intercept0, r_value0, p_value0, std_err0 = stats.linregress(log_N0, log_err0)
slope1, intercept1, r_value1, p_value1, std_err1 = stats.linregress(log_N1, log_err1)

# Create plot
plt.figure(figsize=(10, 6))
plt.scatter(log_N0, log_err0, label='α = 0', color='blue')
plt.scatter(log_N1, log_err1, label='α = 1', color='red')

# Add regression lines
x0 = np.array([log_N0.min(), log_N0.max()])
x1 = np.array([log_N1.min(), log_N1.max()])
plt.plot(x0, slope0 * x0 + intercept0, '--', color='blue', 
         label=f'Régression α=0 (β≈{slope0:.2f})')
plt.plot(x1, slope1 * x1 + intercept1, '--', color='red', 
         label=f'Régression α=1 (β≈{slope1:.2f})')

plt.xlabel('log(N)')
plt.ylabel('log(erreur)')
plt.title('Taux de convergence')
plt.grid(True)
plt.legend()
plt.tight_layout()

print(f"Taux de convergence (β) pour α=0: {slope0:.3f}")
print(f"Taux de convergence (β) pour α=1: {slope1:.3f}")
print(f"Coefficient R² pour α=0: {r_value0**2:.3f}")
print(f"Coefficient R² pour α=1: {r_value1**2:.3f}")
```

# TP4 - EXERCICE 2 - Question 5
# Représentation Graphique

```python
# Nouvelle données pour les schémas explicite et implicite
N_expl = np.array([4, 8, 16, 32, 64, 128, 256])
err_expl = np.array([0.006368028398757719, 0.00155237112702844, 0.0003856365841122811,
                     9.738944401571636e-05, 2.433702968829277e-05, 6.084977268616676e-06,
                     1.521414642990937e-06])

N_impl = np.array([4, 8, 16, 32, 64, 128, 256])
err_impl = np.array([0.006487975973358096, 0.001823123616175482, 0.0005795740370296287,
                     0.0002463108557929083, 0.000139508604728722, 7.465905697362496e-05,
                     3.860322065146748e-05])

# Calcul des logarithmes
log_N_expl = np.log(N_expl)
log_err_expl = np.log(err_expl)
log_N_impl = np.log(N_impl)
log_err_impl = np.log(err_impl)

# Régression linéaire
slope_expl, intercept_expl, r_expl, _, _ = stats.linregress(log_N_expl, log_err_expl)
slope_impl, intercept_impl, r_impl, _, _ = stats.linregress(log_N_impl, log_err_impl)

# Création du graphique
plt.figure(figsize=(10, 6))
plt.scatter(log_N_expl, log_err_expl, color='blue', label='Schéma explicite')
plt.scatter(log_N_impl, log_err_impl, color='red', label='Schéma implicite')

# Ajout des droites de régression
x_expl = np.array([log_N_expl.min(), log_N_expl.max()])
x_impl = np.array([log_N_impl.min(), log_N_impl.max()])
plt.plot(x_expl, slope_expl * x_expl + intercept_expl, '--', color='blue',
         label=f'Régression explicite (β≈{slope_expl:.2f})')
plt.plot(x_impl, slope_impl * x_impl + intercept_impl, '--', color='red',
         label=f'Régression implicite (β≈{slope_impl:.2f})')

plt.xlabel('log(N)')
plt.ylabel('log(erreur)')
plt.title('Taux de convergence des schémas explicite et implicite')
plt.grid(True)
plt.legend()
plt.tight_layout()

print(f"Taux de convergence (β) schéma explicite: {slope_expl:.3f}")
print(f"Taux de convergence (β) schéma implicite: {slope_impl:.3f}")
print(f"Coefficient R² explicite: {r_expl**2:.3f}")
print(f"Coefficient R² implicite: {r_impl**2:.3f}")
```

# TP4 - EXERCICE 3 - Question 4 
# Représentation Graphique pour N = 400

```python
# plots_bs_from_csv.py
# Lit u_final.csv et u_surface.csv et produit les graphes demandés (dont 3D thermique)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

# --------- Réglages ---------
U_FINAL_CSV   = "u_final.csv"    # colonnes attendues: x, u_T
U_SURFACE_CSV = "u_surface.csv"  # colonnes attendues: t, x, u
OUT_DIR       = "figs"           # dossier de sortie pour les images
THERMAL_CMAP  = "inferno"        # 'inferno' (par défaut), 'hot', 'magma', ...

# --------- Utils ---------
def payoff(x):
    # K(x)=0.95 x => h(x)=max(x-K(x),0)=0.05 x
    return 0.05 * x

# --------- Lecture des données ---------
def load_data(u_final_csv, u_surface_csv):
    df_final = pd.read_csv(u_final_csv)
    # robustesse: tolère différents noms de colonnes
    # on tente de repérer x et u_T
    x_col = next((c for c in df_final.columns if c.lower().startswith("x")), None)
    u_col = next((c for c in df_final.columns if c.lower().startswith("u")), None)
    if x_col is None or u_col is None:
        raise ValueError("u_final.csv doit contenir les colonnes 'x' et 'u_T' (ou similaire).")
    X = df_final[x_col].to_numpy()
    uT = df_final[u_col].to_numpy()

    df_surf = pd.read_csv(u_surface_csv)
    # repérage colonnes t, x, u
    t_col = next((c for c in df_surf.columns if c.lower().startswith("t")), None)
    xs_col = next((c for c in df_surf.columns if c.lower().startswith("x")), None)
    us_col = next((c for c in df_surf.columns if c.lower().startswith("u")), None)
    if t_col is None or xs_col is None or us_col is None:
        raise ValueError("u_surface.csv doit contenir les colonnes 't', 'x', 'u'.")

    t_vals = df_surf[t_col].to_numpy()
    x_vals = df_surf[xs_col].to_numpy()
    u_vals = df_surf[us_col].to_numpy()

    # reconstruire la grille 2D à partir du format long (t,x,u)
    Ts = np.unique(t_vals)
    Xs = np.unique(x_vals)
    # on suppose que pour chaque t, on a toutes les valeurs de x dans le même ordre
    U = u_vals.reshape(len(Ts), len(Xs))
    return X, uT, Ts, Xs, U

# --------- Tracés ---------
def make_plots(X, uT, Ts, Xs, U, out_dir=OUT_DIR, thermal_cmap=THERMAL_CMAP):
    out = Path(out_dir); out.mkdir(parents=True, exist_ok=True)

    # 1) Payoff h(x)
    plt.figure()
    plt.plot(X, payoff(X), label="Payoff h(x) = 0.05 x")
    plt.xlabel("x"); plt.ylabel("h(x)"); plt.title("Payoff initial")
    plt.grid(True, linestyle="--", alpha=0.4); plt.legend(); plt.tight_layout()
    plt.savefig(out / "payoff.png", dpi=160)

    # 2) u(T,x) (depuis u_final.csv)
    plt.figure()
    plt.plot(X, uT, label="u(T, x)")
    plt.xlabel("x"); plt.ylabel("u(T, x)"); plt.title("Solution au temps final T")
    plt.grid(True, linestyle="--", alpha=0.4); plt.legend(); plt.tight_layout()
    plt.savefig(out / "u_final.png", dpi=160)

    # 3) quelques coupes temporelles depuis u_surface.csv
    plt.figure()
    nb = min(6, len(Ts))
    idxs = np.linspace(0, len(Ts)-1, nb).astype(int)
    for j in idxs:
        plt.plot(Xs, U[j], label=f"t={Ts[j]:.0f}")
    plt.xlabel("x"); plt.ylabel("u(t, x)"); plt.title("Évolution temporelle (coupes)")
    plt.grid(True, linestyle="--", alpha=0.4); plt.legend(); plt.tight_layout()
    plt.savefig(out / "coupes_temporelles.png", dpi=160)

    # 3') courbes combinées payoff et u(T,x)
    plt.figure()
    plt.plot(X, payoff(X), label="Payoff h(x) = 0.05 x")
    plt.plot(X, uT, label="u(T, x)")
    plt.xlabel("x"); plt.ylabel("Valeur"); plt.title("Payoff et solution finale")
    plt.grid(True, linestyle="--", alpha=0.4); plt.legend(); plt
    plt.tight_layout()
    plt.savefig(out / "payoff_vs_u_final.png", dpi=160)
    

    # 4) Surface 3D (couleurs thermiques)
    Tgrid, Xgrid = np.meshgrid(Ts, Xs, indexing="ij")
    fig = plt.figure()
    ax = fig.add_subplot(111, projection="3d")
    surf = ax.plot_surface(Tgrid, Xgrid, U, cmap=thermal_cmap, linewidth=0, antialiased=True)
    ax.set_xlabel("t"); ax.set_ylabel("x"); ax.set_zlabel("u(t,x)")
    ax.set_title("Surface de la solution (schéma implicite)")
    fig.colorbar(surf, shrink=0.7, aspect=18, pad=0.08, label="u")
    plt.tight_layout()
    plt.savefig(out / "surface_3D.png", dpi=180)

    

    print(f"Images enregistrées dans: {out.resolve()}")
    print(" - payoff.png")
    print(" - u_final.png")
    print(" - coupes_temporelles.png")
    print(" - surface_3D.png")





# --------- Main ---------
if __name__ == "__main__":
    X, uT, Ts, Xs, U = load_data(U_FINAL_CSV, U_SURFACE_CSV)
    make_plots(X, uT, Ts, Xs, U)




#carte graphique de chaleur de u(t,x): inferno, hot, magma
plt.figure()
plt.imshow(U, extent=[Xs[0], Xs[-1], Ts[-1], Ts[0]], aspect='auto', cmap=THERMAL_CMAP)
plt.colorbar(label='u(t,x)')
plt.xlabel('x')
plt.ylabel('t')
plt.title('Carte de chaleur de u(t,x)')
plt.savefig(Path(OUT_DIR) / "carte_chaleur.png", dpi=160  )
```
