```python
import matplotlib.pyplot as plt

# Données pour alpha = 0
N_values = [4, 8, 16, 32, 64, 128, 256, 512, 1024, 2048, 4096, 8192, 16384, 32768]
erreur_alpha_0 = [0.661798, 0.434335, 0.763314, 0.928373, 0.981086, 0.995204, 0.998797, 0.999699, 0.999925, 0.999981, 0.999995, 0.999999, 1, 1]

# Données pour alpha = 1
erreur_alpha_1 = [0.633048, 0.426021, 0.758406, 0.926658, 0.980614, 0.995083, 0.998766, 0.999691, 0.999923, 0.999981, 0.999995, 0.999999, 1, 1]

plt.figure(figsize=(10, 6))
plt.plot(N_values, erreur_alpha_0, marker='o', label='alpha = 0')
plt.plot(N_values, erreur_alpha_1, marker='s', label='alpha = 1')
plt.xscale('log')
plt.xlabel('N')
plt.ylabel('Erreur')
plt.title('Erreur en fonction de N pour différents alpha')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
from scipy.stats import linregress

# Transformation en log
log_N = np.log(N_values)
log_erreur_alpha_0 = np.log(erreur_alpha_0)
log_erreur_alpha_1 = np.log(erreur_alpha_1)

# Nuage de points
plt.figure(figsize=(10, 6))
plt.scatter(log_N, log_erreur_alpha_0, label='alpha = 0')
plt.scatter(log_N, log_erreur_alpha_1, label='alpha = 1')

# Régression linéaire
slope_0, intercept_0, _, _, _ = linregress(log_N, log_erreur_alpha_0)
slope_1, intercept_1, _, _, _ = linregress(log_N, log_erreur_alpha_1)

# Tracer la droite de régression
plt.plot(log_N, slope_0 * log_N + intercept_0, color='blue', linestyle='--', label=f"Régression alpha=0 (β={slope_0:.2f})")
plt.plot(log_N, slope_1 * log_N + intercept_1, color='orange', linestyle='--', label=f"Régression alpha=1 (β={slope_1:.2f})")

plt.xlabel('log(N)')
plt.ylabel('log(erreur_N)')
plt.title('Nuage de points et droite de régression')
plt.legend()
plt.grid(True)
plt.show()

print(f"Taux de convergence expérimental β pour alpha=0 : {slope_0:.2f}")
print(f"Taux de convergence expérimental β pour alpha=1 : {slope_1:.2f}")

# Commentaire :
# La pente β de la droite de régression dans le plan log-log donne le taux de convergence expérimental.
# Plus β est négatif, plus l'erreur décroît rapidement avec N.
```
